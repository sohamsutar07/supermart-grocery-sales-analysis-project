```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
supermart_sales = pd.read_csv(r"D:\data analyst roadmap 1\project guide\supermart grocery project\Supermart Grocery Sales -clean.csv")
```

```python
supermart_sales.groupby('month')['Sales'].sum().plot(kind='line')
plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.show()
```

```python
supermart_sales.groupby('Category')['Sales'].sum().plot(kind='bar')
plt.title("Category Wise Sales")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.show()
```

```python
supermart_sales.groupby('Region')['Sales'].sum().plot(kind='bar')
plt.title("Region Wise Sales")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.show()
```

```python
supermart_sales['Sales'].plot(kind='hist')
plt.title("Sales Distribution")
plt.xlabel("Sales")
plt.show()
```

```python
supermart_sales.plot(x='Sales', y='Profit', kind='scatter')
plt.title("Profit vs Sales")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.show()
```
